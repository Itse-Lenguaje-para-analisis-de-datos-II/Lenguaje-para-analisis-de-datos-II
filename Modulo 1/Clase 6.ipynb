<h1>Estructuras de control</h1>
</div>
En la programación, las estructuras de control son herramientas esenciales para tomar decisiones y controlar el flujo de ejecución de un programa. Python ofrece una variedad de estructuras de control, incluyendo la instrucción if para la toma de decisiones condicionales, los bucles for y while para la repetición de código, y las instrucciones break y continue para controlar el flujo dentro de los bucles.

- IF y ELSE
La instrucción if permite ejecutar diferentes bloques de código en función de una condición que se cumpla o no (verdadera o falsa), lo que se denomina expresión booleana.  

La instrucción else se utiliza como complemento de la instrucción if para crear bloques de código condicional. Su función principal es ejecutar un bloque de código alternativo cuando la condición especificada en la instrucción if se evalúa como False.

- FOR y WHILE
Los bucles for y while permiten repetir un bloque de código un número determinado de veces o mientras se cumpla una condición, respectivamente. 

Los bucles for son útiles para recorrer secuencias de elementos, mientras que los bucles while se utilizan para repetir código hasta que se cumpla una condición específica. 

- BREAK y CONTINUE
Las instrucciones break y continue permiten salir de un bucle antes de que se complete su recorrido normal u omitir el resto del código en la iteración actual y saltar a la siguiente iteración, respectivamente.

La combinación de estas estructuras de control permite a los programadores crear algoritmos más complejos, tomar decisiones inteligentes y resolver problemas diversos de manera eficiente en Python. Desde simples cálculos hasta sofisticados sistemas de inteligencia artificial, las estructuras de control son los cimientos sobre los que se construyen los programas.

## La sentencia if con else
La sentencia if con else en Python se utiliza para realizar una evaluación condicional y ejecutar un bloque de código si la condición especificada es verdadera, pero también ejecutar un bloque de código alternativo si la condición es falsa. La estructura general de la sentencia if con else es la siguiente:

```python
edad = 15
def verificar_edad(edad_jugador):
    if edad >= 16:
        print("El jugador puede participar en la competición.")
    else:
        print("El jugador no cumple con la edad mínima.")

verificar_edad(edad)
```

Analizar la eficacia de los tiros a puerta de los jugadores

```python
def analizar_eficacia(goles, tiros_a_puerta):

  porcentaje_goles = (goles / tiros_a_puerta) * 100

  if porcentaje_goles > 50:

    print("El jugador es muy eficaz en los tiros a puerta.")

  else:

    print("El jugador puede mejorar su eficacia en los tiros a puerta.")


goles_marcados = 20

tiros_a_puerta = 30

analizar_eficacia(goles_marcados, tiros_a_puerta)
```

## La sentencia if con elif
La sentencia elif (abreviatura de else if) se utiliza para agregar múltiples condiciones adicionales después del if y antes del else. Esto permite realizar una serie de comprobaciones condicionales en cascada. La estructura general de la sentencia if con elif y else es la siguiente:
Clasificar equipos por zona de clasificación

```python
def clasificar_equipo(posicion):

  if posicion <= 4:

    print("El equipo clasifica a la Champions League.")

  elif posicion <= 7:

    print("El equipo clasifica a la Europa League.")

  else:

    print("El equipo no clasifica a ninguna competición europea.")


posicion_equipo = 6

clasificar_equipo(posicion_equipo)
```

Calcular bonificación por rendimiento individual y colectivo

```python
def calcular_bonificacion(goles, posicion_equipo):

  if goles > 15 and posicion_equipo <= 3:

    print("Bonificación máxima.")

  elif (goles >= 10 and goles <= 15) or (posicion_equipo >= 4 and posicion_equipo <= 6):

    print("Bonificación media.")

  else:

    print("Bonificación mínima.")

goles_jugador = 12

posicion_equipo = 5

calcular_bonificacion(goles_jugador, posicion_equipo)
```
