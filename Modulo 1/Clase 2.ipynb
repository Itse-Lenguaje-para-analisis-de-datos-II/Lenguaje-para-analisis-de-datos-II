Las variables son uno de los bloques de creación fundamentales de los programas escritos en Python. Son como contenedores que almacenan datos en la memoria de la computadora. Cada variable tiene un nombre único que se utiliza para referirse a su valor almacenado. Los tipos de datos en Python representan el tipo de información que una variable puede contener, como números, cadenas de texto, listas, etc.

### Tipos de variables 
En Python, se declara una variable y se le asigna un valor mediante el operador de asignación =. La variable está en el lado izquierdo del operador y el valor asignado se encuentra en el lado derecho.

- Cadenas de textos (str): Representan un texto normal. Son secuencias de caracteres encerrados entre comillas simples o dobles.

```python
equipo_local = "Real Madrid" 

equipo_visitante = "Barcelona" 

jugador = "Vinicius Jr."
```

Enteros (int): Representan números enteros sin decimales.

```python
goles_local = 2 

goles_visitante = 3 

total_espectadores = 45000 
```

Booleanos (bool): Representan valores verdaderos o falsos. Son útiles para tomar decisiones en estructuras de control.

```python
gano_partido = True 
```

Flotantes (float): Representan números decimales.

```python
puntos_por_partido = 25.5 

rebotes_por_partido = 10.2 

asistencias_por_partido = 6.8
```

## Las funciones int() y float()
En Python, a menudo es necesario convertir cadenas de texto a números para realizar cálculos u otras operaciones. Las funciones int() y float() son herramientas útiles para realizar este tipo de conversiones.

- Convertir a entero (int)
La función int() convierte una cadena de texto en un número entero. Si la cadena contiene un número entero válido, la función devuelve ese número. Si la cadena no contiene un número entero válido, la función generará un error ValueError.

```python
texto_entero = "123"

numero_entero = int(texto_entero)

print(numero_entero)
```

```python
texto_no_entero = "abc" 
try: 
    numero_no_entero = int(texto_no_entero) 

except ValueError: 
    print("Error: La cadena no es un número entero válido") 
```

## Convertir a flotante (float)
La función float() convierte una cadena de texto en un número decimal. Si la cadena contiene un número decimal válido, la función devuelve ese número. 

```python
texto_flotante = "3.14"

numero_flotante = float(texto_flotante)

print(numero_flotante)
```

Si la cadena no contiene un número decimal válido, la función generará un error ValueError. Hay que tener en cuenta que al realizar conversiones de texto a numérico, la cadena sea correcta y esté formateada correctamente para evitar errores. 

```python
texto_no_flotante = "abc" 

try:
    numero_no_flotante = float(texto_no_flotante) 

except ValueError: 
    print("Error: La cadena no es un número decimal válido") 
```

### La función input()
- La función input() en Python es una herramienta fundamental para la interacción entre el programa y el usuario. Se utiliza para obtener datos ingresados por el usuario desde la consola. Permite capturar texto ingresado por el usuario a través de la consola, habilitando una comunicación dinámica y receptiva. Su correcta utilización enriquece la experiencia del usuario y amplía las posibilidades de desarrollo de aplicaciones. 

- La función input() siempre devuelve un valor de tipo str. Si se requiere un tipo numérico, se debe utilizar la función de conversión correspondiente (por ejemplo, int(), float()).

### Funcionamiento 
Los pasos que sigue esta función son:

- Mensaje de solicitud (opcional): 
La función puede mostrar un mensaje en pantalla para guiar al usuario sobre el tipo de información que se espera (por ejemplo, "Ingrese su nombre: ").

- Entrada del usuario: El programa se detiene y espera a que el usuario escriba un valor y presione la tecla Enter.

- Captura y retorno: La función captura el texto ingresado por el usuario y lo devuelve como un valor de tipo cadena de texto (str).

## Ejemplo
En este ejemplo, utilizaremos la función input() para solicitar al usuario que ingrese los datos de un jugador, como su nombre, equipo y posición, entre otros. Comenzaremos pidiendo cada uno de estos datos y almacenándolos en variables correspondientes. En el caso de la edad, que es un valor numérico, usaremos la función int() para convertir el valor ingresado (que inicialmente es una cadena de texto) a un número entero. Finalmente, emplearemos print() para mostrar en pantalla todos los datos del jugador en un formato legible y organizado.

```python
nombre = input("Ingrese el nombre completo del jugador: ")

equipo = input("Ingrese el equipo actual: ")

posicion = input("Ingrese la posición del jugador: ")

numero_camiseta = input("Ingrese el número de camiseta: ")

edad = int(input("Ingrese la edad del jugador: "))

nacionalidad = input("Ingrese la nacionalidad del jugador: ")

print("\nPerfil del jugador:")

print(f"Nombre: {nombre}")

print(f"Equipo: {equipo}")

print(f"Posición: {posicion}")

print(f"Número de camiseta: {numero_camiseta}")

print(f"Edad: {edad} años")

print(f"Nacionalidad: {nacionalidad}")
```

## Aplicaciones 
La función input() cuenta con aplicaciones tan variadas como:

- Recopilación de datos: Se utiliza para obtener información del usuario, como nombres, números, opciones de menú u otro tipo de datos específicos para realizar cálculos o configuraciones. Un ejemplo de uso puede ser un formulario de registro o inscripción. 

- Validación de datos: Permite verificar que el usuario ingresa datos válidos y del tipo correcto (números, fechas, etc.). Un ejemplo de uso puede ser una verificación de formato de un correo electrónico. 

- Interfaz de usuario: Es fundamental para crear interfaces de usuario interactivas en la consola, permitiendo la comunicación entre el programa y el usuario. Un ejemplo de uso puede ser la creación de menús principales de programas. 

- Aplicaciones diversas: Se utiliza en diversos tipos de programas, como juegos, calculadoras, herramientas de configuración, simuladores, etc.
