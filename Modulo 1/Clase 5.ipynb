Al igual que las listas, los diccionarios y las tuplas son estructuras de datos fundamentales en Python que permiten organizar y gestionar información de manera eficiente. Los diccionarios son expertos en la organización por clave-valor, ya que almacenan pares de datos asociados entre sí facilitando el acceso a información específica. Por su parte, las tuplas guardan colecciones ordenadas de elementos que no se pueden modificar una vez creadas, brindando seguridad y confiabilidad en escenarios donde la información debe permanecer intacta.  

## Los diccionarios
Los diccionarios en Python son una estructura de datos que permite almacenar y organizar datos de manera asociativa, es decir, en forma de pares clave-valor. Cada elemento en un diccionario se compone de dos partes: 

- Una clave única que actúa como identificador.

- Un valor asociado a esa clave.

- La clave debe ser inmutable, lo que significa que puede ser una cadena, un número, una tupla u otro tipo de dato que no pueda cambiar. Los valores, en cambio, pueden ser de cualquier tipo, como números, cadenas, listas, tuplas, e incluso, otros diccionarios.

- Los diccionarios se definen utilizando llaves { } y los pares clave-valor se separan por dos puntos :. Los elementos del diccionario están separados por comas. 

- Operaciones básicas con diccionarios
Los diccionarios ofrecen una amplia gama de operaciones que permiten acceder, modificar y manipular sus elementos. Algunas de las operaciones más comunes para crear y manipular un diccionario en Python incluyen: 

Crear un diccionario: Se utiliza la notación {} con pares clave-valor separados por comas y dos puntos.

```python
jugador = {

    "Nombre": "Gerard Moreno",

    "Edad": 30,

    "Equipo": "Villarreal CF",

    "Posición": "Delantero",

    "Goles Marcados": 18

}

for clave, valor in jugador.items():

    print(f"{clave}: {valor}")
```

Acceder a valores por clave:

Se utiliza la clave entre corchetes. 

```python
jugador = {

    "Nombre": "Gerard Moreno",

    "Equipo": "Villarreal CF",

    "Posición": "Delantero",

    "Goles Marcados": 20

}

nombre = jugador["Nombre"]

equipo = jugador["Equipo"]

goles = jugador["Goles Marcados"]

print(f"Nombre del jugador: {nombre}")

print(f"Equipo: {equipo}")

print(f"Goles Marcados: {goles}")
```

Modificar valores en diccionario:

Se asigna un nuevo valor a la clave existente.

```python
jugador["Edad"] = 31

jugador["Goles Marcados"] = 20

print(jugador)
```

Agregar nuevos valores al diccionario:

 Se utiliza la asignación como en la creación, pero con una clave nueva. 

```python
jugador["Asistencias"] = 5

jugador["Partidos Jugados"] = 30

print(jugador)
```

Eliminar elementos del diccionario:

Se utiliza la función del con la clave. 

```python
jugador = {

    "Nombre": "Gerard Moreno",

    "Edad": 31,

    "Equipo": "Villarreal CF",

    "Posición": "Delantero",

    "Goles Marcados": 20

}

del jugador["Edad"]

print(jugador)
```

Verificar si una clave existe en el diccionario:

Se utiliza el operador in

```python
existe_nombre = "nombre" in jugador

existe_goles = "goles" in jugador

print(existe_nombre)   

print(existe_goles)
```

Los diccionarios son herramientas muy útiles cuando necesitas almacenar datos relacionados por una clave única y acceder a ellos rápidamente.  Pueden ser utilizados para representar bases de datos pequeñas, configuraciones, datos estructurados y mucho más. La flexibilidad y eficiencia de los diccionarios los convierten en una poderosa herramienta en Python.

## Las tuplas 
En Python, las tuplas son estructuras de datos similares a las listas, pero con la diferencia principal de que son inmutables. Esto significa que una vez que se crea una tupla, no se pueden modificar sus elementos ni agregar nuevos elementos. Las tuplas se utilizan para almacenar un conjunto de valores relacionados que no deben cambiar.

Las tuplas se definen utilizando paréntesis () y los elementos dentro de la tupla están separados por comas. 

Operaciones básicas con tuplas 
Las tuplas ofrecen una amplia gama de operaciones que permiten acceder, modificar y manipular sus elementos. Algunas de las operaciones más comunes para crear y trabajar con tuplas en Python incluyen: 

Crear una tupla:

Se utilizan paréntesis () para crear una tupla. Los elementos se separan por comas.

```python
mi_tupla = (1, 2, 3, 4, 5)
```

Acceder a elementos de una tupla por índice:

Al igual que las listas, se utiliza el índice entre corchetes para acceder a elementos específicos.

```python
mi_tupla = (10, 20, 30, 40, 50)

primer_elemento = mi_tupla[0]   

ultimo_elemento = mi_tupla[-1]
```

Rebanar una tupla (slicing):

El slicing permite extraer sub-tuplas de la tupla original. Se utiliza la sintaxis tupla[inicio:fin:paso] donde: 

inicio: índice inicial (inclusivo) del elemento a extraer.

fin: índice final (exclusivo) del elemento a extraer.

paso: define el salto entre elementos a extraer (opcional, por defecto es 1).

```python
mi_tupla = (10, 20, 30, 40, 50)

subtupla = mi_tupla[1:4]
subtupla
```

Longitud de una tupla (len):

La función len devuelve la cantidad de elementos en la tupla.

```python
mi_tupla = (1, 2, 3, 4, 5)

longitud = len(mi_tupla)

print(longitud)
```
