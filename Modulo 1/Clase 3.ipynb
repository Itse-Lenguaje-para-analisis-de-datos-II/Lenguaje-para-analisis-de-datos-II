<h1>Los objetos en Python</h1>
</div>
En programación, los objetos son entidades fundamentales que representan datos y el comportamiento asociado a esos datos. A diferencia de los tipos de datos simples, como números o cadenas de texto, los objetos encapsulan datos y las operaciones que se pueden realizar sobre ellos. Esto permite una organización modular y reutilizable del código, lo que facilita la creación y el mantenimiento de programas complejos.

## Python, un lenguaje orientado a objetos 
Python es un lenguaje de programación orientado a objetos, lo que significa que casi todo en Python se implementa como un objeto o una instancia de una clase. Las clases son como plantillas para crear objetos, definiendo sus atributos (características) y métodos (comportamiento).

## Ventajas
La programación orientada a objetos ofrece algunas ventajas significativas:

- Modularidad: Permite dividir el programa en módulos independientes y reutilizables.

- Reusabilidad: Facilita el uso y la adaptación de código existente en nuevos proyectos.

- Abstracción: Oculta los detalles de implementación, permitiendo enfocarse en la lógica del programa.

- Polimorfismo: Permite que diferentes objetos respondan al mismo mensaje de manera diferente.

## Tipos de objetos 
Python ofrece una variedad de tipos de objetos para representar diferentes tipos de datos y estructuras:

- Cadenas de texto: 
Representan secuencias de caracteres alfanuméricos y símbolos. Se pueden crear usando comillas simples o dobles.

- Listas: 
Representan colecciones ordenadas de objetos. Pueden almacenar elementos de diferentes tipos. Se crean usando corchetes.

- Diccionarios: 
Representan colecciones no ordenadas de pares clave-valor. Las claves deben ser únicas y los valores pueden ser de cualquier tipo. Se crean usando llaves.

- Tuplas: 
Representan colecciones ordenadas e inmutables de elementos. No se pueden modificar una vez creadas. Se crean usando paréntesis.

## Las cadenas

Las cadenas de texto, o "strings" en inglés, son uno de los tipos de datos más fundamentales y versátiles en Python. Representan secuencias de caracteres alfanuméricos y símbolos, y se utilizan en una amplia gama de aplicaciones, desde el procesamiento de texto básico hasta la creación de interfaces de usuario sofisticadas.

Las cadenas de texto en Python van más allá de la simple agrupación de caracteres. Son objetos poderosos que poseen atributos y métodos asociados que permiten acceder a sus propiedades y realizar operaciones sobre ellas.

## Crear una cadena:

Las cadenas de texto se pueden crear utilizando comillas simples (') o dobles ("). El contenido entre las comillas representa la secuencia de caracteres que conforma la cadena.

```python
nombre_usuario = "Juan Pérez" 

deporte = "fútbol"

print(f"¡Hola, { nombre_usuario}! Bienvenido al mundo del {deporte}.")
```

## Longitud de una cadena (len):

La longitud de una cadena se refiere a la cantidad de caracteres que la componen. Para obtener la longitud de una cadena, se utiliza la función len().

```python
tweet = "¡Qué partido más emocionante! El Real Madrid ha ganado 3-2 al Barcelona."

longitud_tweet = len(tweet)

print(f"El tweet tiene {longitud_tweet} caracteres.")
```

## Rebanar una cadena (slicing):

El "slicing" o rebanado de cadenas permite extraer subcadenas a partir de una cadena original. Se utiliza una notación similar a la de los rangos en listas, indicando los índices inicial y final (con opción de incluir o excluir el último carácter).

Por ejemplo, cuentas con un registro de los minutos en los que se marcaron los goles en un partido de fútbol. Este registro está almacenado en una cadena, donde cada número representa el minuto en el que se marcó un gol.

```python
goles = "5 12 45 78"

goles_primera_mitad = goles[:4]

cantidad_goles_primera_mitad = len(goles_primera_mitad.split())

print(f"En la primera mitad se marcaron {cantidad_goles_primera_mitad} goles.")
```

## Acceder a caracteres por índice:

Se puede acceder a un carácter individual de una cadena utilizando su índice. Los índices comienzan en 0, y se puede acceder al último carácter utilizando la longitud de la cadena menos 1.

Por ejemplo, tienes el nombre completo de un jugador de fútbol almacenado en una cadena. Queremos extraer solo las iniciales de este jugador (la primera letra de cada nombre). A continuación, se expone una manera de acceder a caracteres por índice: 

```python
nombre_jugador = "Lionel Andrés Messi"

inicial_nombre = nombre_jugador[0]

inicial_apellido = nombre_jugador[7]

iniciales = inicial_nombre + inicial_apellido

print(f"Las iniciales de {nombre_jugador} son {iniciales}.")
```

## Concatenar cadenas:

La concatenación de cadenas permite unir dos o más cadenas para formar una nueva cadena más larga. Se utiliza el operador "+".

```python
equipo_local = "Real Madrid"

equipo_visitante = "Barcelona"

resultado = "2-1"
mensaje_resumen = equipo_local + " " + resultado + " " + equipo_visitante
print(mensaje_resumen)  # Salida: Real Madrid 2-1 Barcelon
```

## Métodos de cadenas:

Las cadenas de texto en Python ofrecen una amplia gama de métodos predefinidos que permiten realizar diversas operaciones sobre ellas. 

```python
nombres_jugadores = ["lIonEL mEsSi", "kRiStIaNo rONaLdO", "nEYmAR jR", "antOInE grIeZmAnN"]

for nombre in nombres_jugadores:

    nombre_limpio = nombre.title().strip()  

    nombre_limpio = nombre_limpio.replace("Jr", "Junior")

    print(nombre_limpio)
```

## Buscar subcadenas (find, index):

El método find() permite buscar la primera aparición de un carácter o subcadena dentro de una cadena. Devuelve la posición del primer carácter coincidente, o -1 si no se encuentra.

```python
linea_jugador = "Luka Modric, Mediocampista"

posicion_mediocampista_find = linea_jugador.find("Mediocampista")

try:

    posicion_mediocampista_index = linea_jugador.index("Mediocampista")

except ValueError:

    posicion_mediocampista_index = -1

if posicion_mediocampista_find != -1:

    print("Usando find(): El jugador es mediocampista.")

else:

    print("Usando find(): El jugador no es mediocampista.")


if posicion_mediocampista_index != -1:

    print("Usando index(): El jugador es mediocampista.")

else:

    print("Usando index(): El jugador no es mediocampista.")
```

## Dividir una cadena (split):

El método split() permite dividir una cadena en una lista de subcadenas según un delimitador especificado.

```python
linea_estadisticas = "Real Madrid, 3, 2, 1, Barcelona, 2, 1, 0"

datos_partido = linea_estadisticas.split(",")

equipo_a = datos_partido[0]

goles_a_favor_a = int(datos_partido[1])

goles_en_contra_a = int(datos_partido[2])

tarjetas_amarillas_a = int(datos_partido[3])

equipo_b = datos_partido[4]

goles_a_favor_b = int(datos_partido[5])

goles_en_contra_b = int(datos_partido[6])

tarjetas_amarillas_b = int(datos_partido[7])


print(f"El {equipo_a} marcó {goles_a_favor_a} goles, recibió {goles_en_contra_a} y tuvo {tarjetas_amarillas_a} tarjetas amarillas.")

print(f"El {equipo_b} marcó {goles_a_favor_b} goles, recibió {goles_en_contra_b} y tuvo {tarjetas_amarillas_b} tarjetas amarillas.")
```

## Comprobar si una cadena comienza o termina con cierto texto (startswith, endswith):

Los métodos startswith() y endswith() permiten verificar si una cadena comienza o termina con un texto específico, respectivamente. Devuelven True si se cumple la condición, y False en caso contrario.

```python
def validar_nombre(nombre):

  return nombre.startswith(nombre[0].upper()) and nombre.endswith('.')


nombres_jugadores = ["Lionel Messi.", "Cristiano Ronaldo Jr.", "Neymar da Silva Santos Júnior.", "Luis Suárez", "Sergio Ramos García."]

contador = 0


for nombre in nombres_jugadores:

  if validar_nombre(nombre):

    contador += 1
print(f"Hay {contador} nombres de jugadores que cumplen con el formato establecido.")
```
