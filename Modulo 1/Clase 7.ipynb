<h1>El bucle for</h1>
</div>
En Python, los bucles for se utilizan para iterar sobre una secuencia (como listas, tuplas, diccionarios, cadenas, etc.) y realizar una acción en cada elemento de la secuencia. La estructura general de un bucle for en Python es la siguiente:

for elemento in secuencia:

Donde:

"elemento" es una variable que tomará el valor de cada elemento de la secuencia en cada iteración.

"secuencia" es la colección de elementos sobre la cual se iterará, por ejemplo, una lista o una cadena.

Iterando sobre una lista

```python
goles_por_partido = [2, 0, 1, 3, 2, 1, 0, 2]
total_goles = 0
cantidad_partidos = len(goles_por_partido)
for goles in goles_por_partido:

    total_goles += goles

promedio_goles = total_goles / cantidad_partidos
print("El promedio de goles por partido es:", promedio_goles)
```

Iterando sobre una cadena

```python
frase_deportiva = "Golear es lo más bonito del fútbol"

for letra in frase_deportiva:

    print(letra)
```

Usando la función range()

```python
num_minutos = 4 

velocidad_por_minuto = [180, 175, 182, 178, 185]  

distancia_total = 0

for i in range(num_minutos):

    distancia_total += velocidad_por_minuto[i]

print(f"El corredor recorrió un total de {distancia_total} metros.")
```

## La función range
La función range es una función incorporada en Python que genera una secuencia de números dentro de un rango especificado. Esta secuencia es útil para crear bucles, iterar sobre elementos o generar listas de números de manera eficiente.

La sintaxis básica de la función range() es la siguiente:

range(start, stop, step)

Donde:

- start: 
El valor inicial del rango (opcional, por defecto es 0).

- stop: 
El valor final del rango. El rango se genera hasta este valor, pero sin incluirlo.

- step: 
El incremento entre los valores (opcional, por defecto es 1).

```python
for num in range(10):

    print(num)
```

```python
for num in range(0, 21, 2):

    print(num)
```

```python
lista_impares = list(range(1, 10, 2))

print(lista_impares)
```

Iterando sobre un diccionario

```python
persona = {

    "nombre": "Juan",

    "edad": 30,

    "ciudad": "Madrid"

}

for clave, valor in persona.items():

    print(f"{clave}: {valor}")
```

Sumar elementos de una lista: calculando el total de puntos anotados por un equipo

```python
puntos_por_partido = [85, 92, 78, 101, 95]
total_puntos = 0

for puntos in puntos_por_partido:

    total_puntos += puntos

print("El equipo anotó un total de", total_puntos, "puntos en la temporada.")
```

Recorrer una cadena y contar 

```python
nombre_jugador = "Karim Benzema"
total_a = 0
for letra in nombre_jugador:
    if letra == 'a' or letra == 'A':

        total_a += 1
print(f"El nombre del jugador Karim Benzema tiene {total_a} 'a'.")
```

Iterar sobre un diccionario que represente a un equipo de fútbol

```python
equipo = {

    "nombre": "Real Madrid",

    "estadio": "Santiago Bernabéu",

    "entrenador": "Carlo Ancelotti",

    "fundado": 1902

}

for clave, valor in equipo.items():

    print(f"{clave}: {valor}")
```

Iterar sobre un diccionario y mostrar sus claves y valores

```python
equipo = {

    "goles_a_favor": 85,

    "goles_en_contra": 40,

    "partidos_ganados": 25,

    "partidos_empatados": 10,

    "partidos_perdidos": 5

}

for estadistica, valor in equipo.items():

    print(f"El equipo ha tenido {valor} {estadistica}.")
```
