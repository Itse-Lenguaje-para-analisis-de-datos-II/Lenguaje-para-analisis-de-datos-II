<h1>El bucle while</h1>
</div>
El bucle while en Python es una estructura de control que se utiliza para repetir un bloque de código mientras una condición específica sea verdadera. En otras palabras, el bucle while ejecutará el bloque de código repetidamente hasta que la condición especificada sea falsa.

Donde:

condición es una expresión booleana que se evalúa en cada iteración del bucle.

Si la condición es verdadera, el bloque de código dentro del bucle se ejecutará.

Si la condición es falsa, el bucle while se detendrá y la ejecución continuará con la siguiente instrucción después del bucle.

Ejemplos 

Contador regresivo

```python
contador = 5
while contador > 0:
    print(contador)
    contador -= 1
```

En este ejemplo, el bucle while imprimirá el valor del contador en cada iteración mientras sea mayor que cero. El contador comenzará con el valor 5, y en cada iteración, se imprimirá el valor actual y se le restará 1 hasta que el contador llegue a cero y la condición sea falsa, momento en el que el bucle se detendrá

Suma de números positivos

```python
suma = 0
while True:
    numero = int(input("Ingrese un número positivo (ingrese 0 para detener): "))
    if numero <= 0:
        break
    suma += numero
print("La suma de los números positivos ingresados es:", suma)
```

Imprimir sólo números pares

```python
numero = 1
while numero <= 10:
    if numero % 2 != 0:
        numero += 1
        continue
    print(numero)
    numero += 1
```
