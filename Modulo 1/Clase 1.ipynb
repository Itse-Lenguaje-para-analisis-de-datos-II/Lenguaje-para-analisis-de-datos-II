### Instalación de Python 
Lo primero para empezar a programar en Python es instalar el programa en nuestro sistema operativo. A continuación, veremos los pasos a seguir: 

- Visita el sitio web oficial: 
Accede al sitio web oficial de Python en https://www.python.org/ y navega hasta la sección de descargas.


- Descarga el instalador: 
En la sección de descargas, encontrarás diferentes versiones de Python disponibles. Siempre es recomendable descargar la última versión estable. Elige el instalador adecuado para tu sistema operativo (Windows, macOS o Linux).

- Ejecuta el instalador: 
Una vez descargado el instalador, ábrelo y ejecútalo como administrador en caso de ser necesario.

- Opciones de instalación: 
Durante el proceso de instalación, es posible que se te presenten algunas opciones adicionales. Asegúrate de seleccionar "Add Python X.X to PATH" (donde "X.X" es la versión de Python que estás instalando). Esta opción añade Python al PATH del sistema, lo que permitirá acceder a Python desde cualquier ubicación en la línea de comandos.


- Completa la instalación: 
Continúa con el proceso de instalación siguiendo las instrucciones en pantalla. Python se instalará en tu ordenador en unos minutos.

- Verifica la instalación: 
Una vez finalizada la instalación, abre la línea de comandos o terminal y escribe "python" seguido de "Enter". Deberías ver el intérprete de Python mostrando la versión que has instalado. También puedes escribir "python --version" para obtener la versión instalada.



Una vez tengas Python instalado en tu ordenador ya estarás listo para comenzar a programar en este lenguaje. Puedes escribir y ejecutar programas Python desde la línea de comandos, o utilizar un entorno de desarrollo integrado (IDE) como PyCharm, Visual Studio Code o IDLE para una experiencia más completa y amigable. ### Entorno de desarrollo integrado (IDE)
Un IDE o entorno de desarrollo integrado es una herramienta que facilita la escritura, edición y ejecución de código, mejorando la experiencia de programación por las diversas funcionalidades que proporciona. Algunas de las características que tienen en común estas herramientas son:

- Editor de código: Permite escribir y editar código de manera cómoda, con resaltado de sintaxis, autocompletado de código y otras funciones útiles para el programador.

- Compilador o intérprete: Ejecuta el código escrito y muestra los resultados o errores.

- Depurador: Permite identificar y corregir errores en el código.

- Herramientas de administración de proyectos: Facilita la organización y gestión de proyectos de programación complejos.

- Integración con herramientas de control de versiones: Permite realizar un seguimiento de los cambios realizados en el código y colaborar con otros desarrolladores. 

### Las variables en Python 

Las variables son uno de los bloques de creación fundamentales de los programas escritos en Python. Son como contenedores que almacenan datos en la memoria de la computadora. Cada variable tiene un nombre único que se utiliza para referirse a su valor almacenado. Los tipos de datos en Python representan el tipo de información que una variable puede contener, como números, cadenas de texto, listas, etc.
