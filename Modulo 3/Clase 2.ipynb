La librería datetime proporciona clases y funciones para trabajar con fechas, horas y duraciones de tiempo. Permite manejar información de fechas y horas de manera sencilla y precisa, lo que lo convierte en una herramienta útil para tareas relacionadas con el manejo del tiempo y los calendarios.

¡Recuerda!
Las librerías o bibliotecas son herramientas adicionales que expanden las capacidades de un lenguaje de programación, en este caso Python. Son conjuntos de código ya escrito que ofrecen al programador una amplia variedad de funcionalidades para realizar tareas específicas, ahorrando tanto tiempo como esfuerzo. Las librerías permiten analizar, manipular y visualizar datos, desarrollar aplicaciones web, automatizar tareas o acceder a APIs.

Dentro de datetime, podemos encontrar algunas clases importantes:

- datetime: Representa una combinación de fecha y hora.

- date: Representa una fecha (año, mes y día) sin información de hora.

- time: Representa una hora del día (horas, minutos, segundos y microsegundos).

- timedelta: Representa una duración o diferencia entre dos puntos de tiempo.

- tzinfo: Representa información sobre la zona horaria.


```python
from datetime import datetime, date, time
```

```python
fecha_hora = datetime(2023, 8, 1, 12, 30, 0)

print(fecha_hora)  
```

## Objeto date que sólo contenga la fecha

```python
fecha = date(2023, 8, 1)

print(fecha)  
```

## Objeto time que sólo tenga la hora

```python
hora = time(12, 30, 0)

print(hora) 
```

## Obtención de fecha y hora

```python
from datetime import datetime

fecha_hora_actual = datetime.now()

print(fecha_hora_actual) 
```

## Operaciones con timedelta: Podemos obtener una nueva fecha basándonos en la fecha actual.

```python
from datetime import datetime, timedelta

# Obtener la fecha y hora actual

fecha_hora_actual = datetime.now()



# Crear un timedelta de 7 días

duracion = timedelta(days=7)



#Sumar el timedelta de 7 días a la fecha actual para obtener la nueva

nueva_fecha = fecha_hora_actual + duracion

print(nueva_fecha)  
```

## d.strftime
En ocasiones, necesitamos la fecha y la hora en un formato específico y datetime nos permite organizarlas en el que requiramos. El método d.strftime nos devuelve la cadena que resulta de transformar la fecha  con el formato indicado en la cadena formato. 

```python
from datetime import datetime



# Obtener la fecha y hora actual

fecha_hora_actual = datetime.now()



# Formatear la fecha y hora en el formato que especifiquemos. En este caso, hemos elegido año-mes-día hora-minuto-segundo

formato = "%Y-%m-%d %H:%M:%S"

fecha_hora_formateada = fecha_hora_actual.strftime(formato)

print(fecha_hora_formateada)  
```

Además de los marcadores de posición que hemos empleado en el ejemplo anterior, la cadena formato puede contener como: 

- %Y (año completo)

- %y (últimos dos dígitos del año)

- %m (mes en número)

- %B (mes en palabra)

- %d (día)

- %A (día de la semana)

- %a (día de la semana abreviado)

- %H (hora en formato 24 horas)

- %I (hora en formato 12 horas)

- %M (minutos)

- %S (segundos)

- %p (AM o PM)

- %C (fecha y hora completas)

- %x (fecha completa)

- %X (hora completa)
