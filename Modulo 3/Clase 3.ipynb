# Plotnine

El paquete Plotnine, que forma parte de "La gramática de gráficos", se usa para la creación de gráficos complejos a partir de los datos en un DataFrame. Utiliza configuraciones por defecto que ayudan a crear gráficos con calidad de publicación con pocos ajustes.

## Instalación de Plotnine
La instalación de Plotnine es un proceso sencillo que te permitirá comenzar a crear visualizaciones de datos atractivas y reveladoras. A continuación, puedes ver el código para instalar Plotnine de dos maneras: utilizando conda y pip.

## Importación de la librería 
La instrucción from plotnine import* importa todos los objetos y funciones de la librería Plotnine en el espacio de nombres global. Si bien esto puede ser útil en algunos casos, se recomienda utilizar importaciones específicas para evitar conflictos de nombres y mejorar la legibilidad del código.

```python
from plotnine import*
import  seaborn as sns
```

Por ejemplo, para importar  solo la función ggplot(), utiliza from plotnine import ggplot. Para importar todas las funciones y objetos esenciales, utiliza import plotnine as pn.
## Sintaxis base
La base de cualquier gráfico creado con Plotnine es la siguiente:

```python
df=sns.load_dataset("tips")
```

```python
df.info()
```

```python
grafica = (
    ggplot(df, aes(x="total_bill", y="tip"))
    + geom_point()
)
grafica
```

```python
ggplot(df,aes(x="total_bill" , y="tip"))+ geom_point()
```

Dentro de esta línea de código, df se refiere al Dataframe del que extraigamos los datos, mientras que col1 y col2 son las variables a graficar. Después de escribr esos datos, agregaremos el tipo de gráfico que queremos emplear, como por ejemplo un gráfico de barras, un histograma o un boxplot, entre otros:

```python
ggplot(aes(x="sex" , y="total_bill"),df) + geom_bar(stat="identity")+xlab ("Texto en el eje x")+ylab ("Texto en el eje y")+ggtitle ("Título del gráfico")

```
