<h1>Heatmaps Plotnine</h1>
</div>
Un heatmap o mapa de calor es una representación gráfica de datos en una matriz cuyos valores individuales se representan como colores. Este tipo de gráfico es útil para visualizar la densidad de los datos, la correlación entre variables o realizar un análisis de patrones.

```python
import pandas as pd
from plotnine import*
from plotnine.data import *
```

```python
data=pd.read_excel(r"C:\Users\Easy Data\Dropbox\PC\Documents\Lenguajes para analisis de datos II\Materiales\bar.xlsx")
```

```python
ggplot(aes("week_day","month", fill="bill"),data)+geom_tile()
```

Para poder separar las celdas y facilitar la separación de colores adyacentes similares, añadimos comandos dentro del geom_tile().

```python
ggplot(aes("week_day","month", fill="bill"),data) +geom_tile(color="white",size=1)
```

Usamos scale_fill_gradient() para cambiar los colores del gráfico. 

```python
ggplot(aes("week_day","month", fill="bill"),data) +\
geom_tile(color="white",size=1) +\
scale_fill_gradient(low="lightblue", high="Blue")
```

```python
ggplot(aes("week_day","month", fill="bill"),data) +\
geom_tile(color="white",size=1) +\
scale_fill_gradient(low="#F492CF", high="red")
```

Para finalizar el gráfico, añadimeros el título, el bombre a los ejes y un theme().

```python
ggplot(aes("factor(week_day)","factor(month)", fill="bill"),data) +\
geom_tile(color="white",size=1) +\
scale_fill_gradient(low="#F492CF", high="red") +\
ggtitle("Propinas por semanas y mes") +\
xlab("Dias de la semana")+ ylab("Meses")+ theme_classic()
```

Segundo ejemplo de creación de heatmap
Con el mismo dataset que hemos usado en el primer ejemplo, vamos a separar las propinas por nacionalidades y meses en vez de días de la semana y meses.

1. Seleccionamos la información que queremos usar del dataset

```python
data2 = data.loc[:,["month","nationality","tip"]].groupby(by=["month","nationality"],as_index=False).sum()
```

Creamos un gráfico con la información que hemos seleccionado. Al igual que hemos hecho antes, le añadimos las líneas entre las celdas, elegimos la paleta de color que queremos emplear y añadimos los nombres de los ejes y del gráfico.

```python
ggplot(aes("factor(month)","nationality", fill="tip"),data2) +\
geom_tile(color="white", size=1) +\
scale_fill_gradient(low="#FFA533", high="yellow") +\
ggtitle("Propinas mensuales por meses")+xlab("Meses") +\
ylab("Nacionalidades") +theme_classic()
```

Añadimos el valor en cada cuadro con geom_text().

```python
ggplot(aes("factor(month)","nationality", fill="tip"),data2) +\
geom_tile(color="white", size=1) +\
scale_fill_gradient(low="#FFA533", high="yellow") +\
ggtitle("Propinas mensuales por meses")+xlab("Meses") +\
ylab("Nacionalidades") +theme_classic() +\
geom_text(aes(label="tip"),size=8, color="black")
```
