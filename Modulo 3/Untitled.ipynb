```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from plotnine import *
```

```python
iris=sns.load_dataset("iris")
iris
```

```python
iris.info()
```

```python
iris["sepal_length"].plot(kind="hist")
```

```python
plt.hist(iris["sepal_length"],bins=5)
```

```python
sns.histplot(iris["sepal_length"],kde=True,bins=4)
```

```python
ggplot(aes("sepal_length"),iris)+geom_histogram()
```

```python
iris["species"].value_counts().plot(kind="bar")
```

```python
data=iris["species"].value_counts()
plt.bar(x=data.index,height=data.values)
plt.grid(True)
```

```python
sns.barplot(x=iris["species"].value_counts().index,y=iris["species"].value_counts().values)
```

```python
sns.countplot(data=iris,x="species")
```

```python
iris["species"].value_counts().plot(kind="pie",autopct="%2.2f")
```

```python
plt.pie(data=iris,x=iris["species"].value_counts(),autopct="%2.1f",labels=iris["species"].value_counts().index,explode=[0,0.2,0.4],shadow=True)
```

```python
iris.groupby(by="species").size().plot(kind="pie",autopct="%2.1f")
```

```python
sns.pairplot(iris)
```

```python
iris["sepal_length"].plot(kind="box")
```

```python
plt.boxplot(iris["sepal_length"])
```
