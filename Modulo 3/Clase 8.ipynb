Matplotlib es una librería de Python especializada en la creación de gráficos en dos dimensiones. Permite crear y personalizar los tipos de gráficos más comunes. 

Generación de un Dataframe
Para poder trabajar con Matplotlib, vamos a generar nuestro propio Dataframe, que usaremos para crear los gráficos.

1. Generamos los datos en Pandas ejecutando este código:

```python
import pandas as pd
```

Ahora creamos el dataset, que incluirá lo siguiente:

```python
al =["Alemania",4,2014,"Brasil"]

sp =["España",1,2010,"sudafrica"]

bz =["Brasil",5,2002,"Japon"]

it =["Italia",4,2006,"Alemania"]

uk =["Inglaterra",1,1966,"inglaterra"]

uru =["Uruguay",2,1950,"Brasil"]

arg =["Argentina",2,1986,"Mexico"]

fr =["Francia",2,2018,"Rusia"]
```

Creamos la lista de datos:

```python
lista=[sp,bz,it,al,uru,arg,fr,uk]
```

```python
df=pd.DataFrame(lista, columns=["Pais","Copas","Año","Lugar_de_victoria"])
```

## Ejemplo de creación de un gráfico de puntos
Antes de empezar a crear el gráfico, ejecutamos esta línea de código para importar Matplotlib de forma que nos permita usar plt en lugar de matplotlib.pyplot, simplificando el código que escribiremos.

```python
import matplotlib.pyplot as plt
```

Creamos el gráfico con plt.plot() e indicamos los números que aparecerán en el eje x y en el eje y.

```python
x=[1,2,3,4,5]

y=[1,6,35,69,140]

plt.plot(x,y)

plt.show()
```

Usamos Matplotbit para editar el gráfico que acabamos de crear, cambiando las variables de los ejes a Pais y Copas.

```python
plt.plot(df["Pais"],df["Copas"])

plt.show()
```

 Para transformarlo en un scatterplot, usaremos plt.scatter().

```python
plt.scatter(df["Pais"],df["Copas"])

plt.show()
```

Como los nombres de los países están demasiado juntos o incluso superpuestos, vamos a girar el eje X 45 grados con plt.xticks().

```python
plt.scatter(df["Pais"],df["Copas"])

plt.xticks(rotation=45)

plt.show()
```

Ejemplo de creación de un gráfico de barras
1. Vamos a utilizar Pais y Copas para crear un gráfico de barras con plt.bar().

```python
plt.bar(df["Pais"],df["Copas"])

plt.show()
```

Para poder ver los países claramente sin necesidad de rotarlos, vamos a transformarlo en un gráfico horizontal con plt.barh().

```python
plt.barh(df["Pais"],df["Copas"])

plt.show()
```

Si queremos que el gráfico esté en orden ascentente o descendente, necesitamos modificar el Dataframe original con df.sort_values().

```python
df=df.sort_values("Copas",ascending=False)

plt.bar(df["Pais"],df["Copas"])

plt.show()
```

Podemos añadirle colores a las barras con color=[].

```python
df=df.sort_values("Copas",ascending=False)

plt.bar(df["Pais"],df["Copas"],

color=["b","r","g","m","k","c","y","b"])

plt.show()
```

En el caso de Matplotlib, hay unos colores predeterminados disponibles, y cada uno cuenta con un símbolo para identificarse. A continuación puedes ver todos los colores que existen junto a sus símbolos:

Añadimos un título al gráfico con ply.title(), y definimos los nombres de los ejes  con plt.xlabel() y plt.ylabel(). También vamos a aprovechar para rotar los nombres de los países del eje X con plt.xticks().

```python
df=df.sort_values("Copas",ascending=False)

plt.bar(df["Pais"],df["Copas"],

color=["b","r","g","m","k","c","y","b"])

plt.xticks(rotation=45)

plt.xlabel('Paises')

plt.ylabel('Copas Ganadas')

plt.title('Copas ganadas en los Mundiales de Futbol');

plt.show()
```

Creamos una cuadrícula con plt.grid(), que nos puede ayudar a facilitar las comparaciones entre diferentes barras en gráficos más complejos.

```python
df=df.sort_values("Copas",ascending=False)

plt.bar(df["Pais"],df["Copas"],

color=["b","r","g","m","k","c","y","b"])

plt.xticks(rotation=45)

plt.xlabel('Paises')

plt.ylabel('Copas Ganadas')

plt.title('Copas ganadas en los Mundiales de Futbol');

plt.grid(True)

plt.show()
```

## creación de gráficos de pastel
Los gráficos de pastel o pie charts pueden sufrir algunas malinterpretaciones al ser analizados, por lo que tendremos que tener algo de cuidado extra al crearlos.

1. Vamos a utilizar Copas y Pais para crear un gráfico de pastel con plt.pie().

```python
plt.pie(df["Copas"],labels=df["Pais"])

plt.show()
```

2. Podemos mostrar los números dentro de las diferentes rebanadas para reforzar la visualización con autopct. Con este argumento, podemos hacer dos cosas:

Decidir la cantidad de decimales que tendrá cada número.

autopct = '%.1f' # Muestra el valor con un decimal

autopct = '%.2f' # Muestra el valor con dos decimales

Mostrar el símbolo % en el gráfico.

autopct = '%.1f%%'

autopct = '%.2f%%' 

## Extracción de rebanadas de un gráfico de tarta
Los gráficos de tarta permiten segmentar o separar una o más rebanadas del resto con el parámetro explode=[]. Para ello, necesitaremos una lista con valores entre 0 y 1, que indicarán el desfase con respecto al centro del gráfico. 0 indica que no hay desfase y 1 que la rebanada se desplazará una distancia igual al radio del gráfico.

1. Extraemos las rebanadas del gráfico con explode, indicando el desfase para cada una de ellas.

```python
plt.pie(df["Copas"],labels=df["Pais"],autopct="%.2f%%",

explode=[0,0.1,0,0.2,0,0.4,0,0.2])

plt.show()
```

Añadimos una sombra con shadow=True para facilitar la visualización.

```python
plt.pie(df["Copas"],labels=df["Pais"],autopct="%.2f%%",

explode=[0,0.1,0,0.2,0,0.4,0,0.2],shadow=True)

plt.show()
```

Si queremos cambiar los colores, crearemos una lista de los que queremos emplear usando códigos hexadecimales para cada rebanada con colors=[].

```python
colores=['#FFEBCD', '#00FFFF', '#DEB887', '#d62728', '#8A2BE2', '#e377c2', '#5F9EA0', '#A52A2A']

plt.pie(df["Copas"],labels=df["Pais"],

    	colors=colores,

    	shadow=True,

    	explode=[0,0.1,0,0.2,0,0.4,0,0.2],

    	autopct="%.1f%%",

    	)

plt.show()
```
