Los barplots y los density plots son esenciales para comparar categorías y examinar la distribución continua de los datos, ayudando a revelar patrones y tendencias en el análisis de datos.

Barplots
Un barplot o gráfico de barras es una representación gráfica que se emplea para poder visualizar la comparación entre diferentes categorías de datos. Las barras pueden ser verticales u horizontales y su longitud será determinada por el valor que representa.

Los barplots se pueden utilizar, por ejemplo, para comparar las ventas de diferentes productos o las calificaciones de diferentes servicios.  

Ejemplo de creación de un barplot
Para este ejemplo, usaremos el dataset bar.xlsx.

```python
import pandas as pd
from plotnine import*
from plotnine.data import *
```

```python
data=pd.read_excel(r"C:\Users\Easy Data\Dropbox\PC\Documents\Lenguajes para analisis de datos II\Materiales\bar.xlsx")
```

```python
ggplot(aes("service"),data)+geom_bar() 
```

Si queremos hacer el gráfico horizontal sólo tenemos que añadir el comando coord_flip().

```python
ggplot(aes("service"),data)+geom_bar()+coord_flip()
```

Para añadir color al gráfico, usamos fill aesthetics:

```python
data = data.loc[data.service.isin (["brilliant","very good","excellent"]),:]

ggplot(aes(x="service",fill="service"),data)+geom_bar()+coord_flip()
```

Si queremos emplear otros colores que se ajusten más a nuesta temática, podemos elegir entre multitud de paletas que ya están creadas. Si queremos emplearlas, usaremos scale_fill_brewer. A continuación, tienes algunos ejemplos.

```python
ggplot(aes("service", fill="service"),data)+geom_bar()+\
scale_fill_brewer(type = "qual", palette = "Pastel1")
```

```python
ggplot(aes("service", fill="service"),data)+geom_bar()+\
scale_fill_brewer(type = "seq", palette = "PuBu")
```

```python
ggplot(aes("service", fill="service"),data)+geom_bar()+\
scale_fill_brewer(type = "div", palette = "RdGy")
```

## Density plot
Un density plot o gráfico de densidad es una representación gráfica que muestra la distribución de una variable continua suavizada. Este tipo de gráfico es similar a los histogramas, pero obtiene distribuciones más fluidas al suavizar el kernel1 en el momento de representar los valores. Los picos de densidad que aparecen en el gráfico ayudan a visualizar dónde se concentran los valores en el intervalo seleccionado.

Los density plots son útiles para visualizar distribuciones continuas, como por ejemplo, la distribución de ingresos en una población.

1 En este contexto, el término "kernel" se refiere a la función utilizada para suavizar los datos y estimar la densidad de probabilidad. En lugar de simplemente contar las observaciones en cada intervalo como en un histograma, el kernel aplica una función matemática para crear una curva suave que muestra cómo se distribuyen los datos. Esto permite una visualización más fluida y continua de la distribución de una variable.

```python
ggplot(aes("bill"),data) + geom_density()
```

Añadimos fill aesthetics para agregar color y facilitar el visualizado.

```python
ggplot(aes("bill", fill="service"),data) + geom_density()
```
