## Crosstab
La función pandas.crosstab se utiliza para calcular una tabla de frecuencias cruzadas, también conocida como tabla de contingencia, a partir de dos o más variables. Cuenta con utilidad en campos como el de análisis de asociación, donde permite determinar si existe alguna asociación o relación entre las categorías de las variables, y la visualización de datos, que ayuda a comprender mejor la distribución de datos usando gráficos de todo tipo.

Por defecto, el resultado de esta función evalúa las frecuencias, ya sean absolutas o relativas, de cada combinación de valores, pero es posible especificar una función de agregación. Si contamos con una fila Phase y una columna NumberofShells, la función se escribiría así:

```python
pd.crosstab(df.Phase, df.NumberofShells)
```

## Funciones de agregación en crosstab
Existen tres  funciones que podemos agregar a la función crosstab:

Para obtener un resultado que represente el porcentaje del total. Añadiremos normalize="all" tras la segunda variable en la línea de código.

```python
pd.crosstab(df.Phase, df.NumberofShells, normalize="all")
```

Para obtener un resultado que represente el porcentaje por fila. Añadiremos normalize="index" tras la segunda variable en la línea de código.

```python
pd.crosstab(df.Phase, df.Type, normalize="columns")
```

## Combinación de Dataframes
La combinación de dos Dataframes en un único Dataframe se puede realizar de dos maneras: 

- Concatenación de filas: Las filas de los Dataframes se concatenan unas a continuación de las otras para formar el nuevo DataFrame. Para ello, es necesario que los DataFrames que se combinen tengan el mismo índice de columnas.

```python
df1 = pd.DataFrame({"Nombre":["Carmen", "Luis"],"Sexo":["Mujer", "Hombre"], "Edad":[22, 18]}).set_index("Nombre")
df2 = pd.DataFrame({"Nombre":["María", "Pedro"], "Sexo":["Mujer", "Hombre"], "Edad":[25, 30]}).set_index("Nombre")
df = pd.concat([df1, df2])
df.reset_index(inplace=True)
```

## Concatenación de columnas: 
Las columnas de los DataFrames se concatenan unas a continuación de las otras para formar el nuevo DataFrame. Para ello, es necesario que los DataFrames que se combinan tengan el mismo índice de filas.

```python
df1 = pd.DataFrame({"Nombre":["Carmen", "Luis", "María"],"Sexo":["Mujer", "Hombre", "Mujer"]}).set_index("Nombre")
df2 = pd.DataFrame({"Nombre":["Carmen", "Luis", "María"], "Edad":[22, 18, 25]}).set_index("Nombre")

df = pd.concat([df1, df2], axis = 1)

df.reset_index(inplace=True)
```

## Merge

## El método merge() 
En Pandas sirve para combinar los datos de dos Dataframes de forma eficiente. Requiere recibir dos Dataframes como parámetros y las instrucciones para combinarlos. La especificación de este método es la siguiente:

- left: incluye los valores del primer objeto DataFrame y los del segundo que cumplan la condición de unión.

- right: incluye los valores del segundo objeto DataFrame y los del primero que cumplan la condición de unión.

- outer: con esta opción se incluyen tanto los valores del primer como del segundo objeto, aunque no cumplan la condición de unión.

- inner: esta es la opción por defecto y solamente devuelve los registros de ambos objetos DataFrame que cumplan la condición de unión.

- on: es el nombre de la columna de ambos objetos por cuyo valor se unen los objetos DataFrame. En caso de que no se pase ningún valor a Pandas, los objetos se unirán mediante los índices. 

En caso de que el nombre de la columna mediante la que se desea unir los objetos no sea el mismo en ambos Dataframe, se deberán usar los parámetros left_on y right_on para indicar el nombre del primer y segundo objeto.

Ejemplo de combinación de Dataframes
Contamos con 2 conjuntos de datos que puedes encontrar en la carpeta de Recursos del Módulo.

df_1, que muestra el id, nombre, edad y cargo de algunos empleados.

df_2, que muestra el id, nombre y salario mensual de algunos empleados.

```python
total = pd.merge(df1, df2, on=["id","name"],how="inner")
```

```python
total = pd.merge(df1, df2, left_on=["id","name"],right_on=["id","name"],how="left")
```

```python
total = pd.merge(df1, df2, left_on=["id","name"],right_on=["id","name"],how="right")
```

```python
total = pd.merge(df1, df2, on=["id","name"],how="outer")
```
