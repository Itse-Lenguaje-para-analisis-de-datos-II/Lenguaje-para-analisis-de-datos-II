## ¿Qué es Pandas?
En el ámbito de la ciencia de datos, existen numerosas librerías de Python para elegir, pero hay una que destaca en el análisis y manipulación de estructuras de datos: Pandas.

Pandas fue creada para facilitar y mejorar el manejo de datos estructurados, especialmente en forma de tablas y series de tiempo. Proporciona dos estructuras de datos fundamentales que permiten representar y trabajar con datos de manera eficiente: Dataframe y Series. 



## Usos de Pandas
Pandas es ampliamente utilizada en el análisis de datos y la ciencia de datos debido a su versatilidad y funcionalidad. Algunas de las aplicaciones para las que se utiliza son: 

- Carga y manipulación de datos: Pandas permite cargar datos desde diferentes fuentes, como archivos CSV, Excel, bases de datos y más. También proporciona herramientas para limpiar y preprocesar datos, como el manejo de datos faltantes y duplicados.

- Análisis exploratorio de datos: Pandas facilita el análisis y la exploración inicial de los datos, permitiendo realizar consultas, filtrar, ordenar y agregar datos de manera sencilla.

- Operaciones de datos: Pandas proporciona una amplia gama de funciones y métodos para realizar operaciones matemáticas, estadísticas y de manipulación de datos en las estructuras DataFrame y Series.

- Visualización de datos: Pandas se integra perfectamente con librerías de visualización como Matplotlib y Seaborn, lo que permite crear gráficos y visualizaciones de datos de manera sencilla.

- Integración con otras herramientas de ciencia de datos: Pandas se utiliza comúnmente en conjunto con otras librerías de Python, como NumPy, Scikit-learn y TensorFlow, para realizar análisis y modelado de datos más complejos.

Los DataFrames son una de las herramientas fundamentales de Pandas para el análisis de datos. Estas estructuras de datos bidimensionales (similares a las hojas de cálculo) permiten almacenar y organizar grandes cantidades de información en forma de tablas con filas y columnas. Los DataFrames pueden albergar una amplia variedad de tipos de datos, incluyendo caracteres, números enteros, valores de punto flotante, fechas, horas, valores booleanos y factores, entre otros.

5. Obtención de columnas o filas  
En Pandas, la obtención de columnas o filas específicas de un dataset es una operación fundamental, especialmente cuando se trabaja con conjuntos de datos grandes o complejos. Pandas te ofrece dos herramientas para realizar esta selección con precisión: loc e iloc.

## El método loc 
Te permite seleccionar filas o columnas por su nombre o etiqueta. Es ideal cuando conoces los nombres específicos de las filas o columnas que deseas extraer. Su sintaxis es sencilla: df.loc[filas, columnas]. 

## El método iloc
Te permite seleccionar filas o columnas por su índice numérico. Es ideal cuando conoces la posición exacta de las filas o columnas que deseas extraer. Su sintaxis es sencilla: df.iloc[filas, columnas]. 

También puedes combinar ambos métodos para una selección más avanzada. Para seleccionar todas las filas o columnas, emplearás dos puntos :, mientras que emplearás el nombre o el index concreto si necesitas especificar qué columnas o filas que quieres extraer. 

Ejemplos 
Seleccionar columnas y filas por su nombre

En este caso, emplearemos el nombre de las filas o columnas que queremos extraer:

```python
df.loc[:,["Element","Symbol"]]
```

## Seleccionar columnas y filas por index

En el caso de querer usar el index, debes recordar que en Python empieza en 0, no en 1. Al igual que en el ejemplo anterior, si quieres seleccionar todas, usarás dos puntos :.

```python
df.iloc[0,:]
```

```python
df.iloc[:,5:8]
```

```python
df.iloc[2:6,:]
```

## Seleccionar columnas y filas mezclando index y nombres

También se pueden combinar ambos métodos, encadenando .loc e .iloc para hacer la selección de los datos.

```python
df.iloc[1:3, :].loc[:, ["Element", "Symbol"]]
```

```python
df.iloc[0:3, :].iloc[:, 3:6]
```

## Selecciones condicionales
En el análisis de datos con Pandas, la selección de datos no se limita a extraer filas o columnas específicas. Con las selecciones condicionales, puedes obtener solo los datos que cumplen con ciertos criterios, permitiéndote enfocarte en la información que realmente importa para tu análisis.

- Filtrado por condiciones

Para realizar selecciones condicionales, se utiliza una expresión booleana que define la condición que deben cumplir los datos para ser incluidos. Esta expresión se aplica a una columna específica del DataFrame.

```python
df.loc[df.AtomicMass > 280,:]
```

## Combinación de operadores booleanos 

Puedes combinar operadores booleanos (&, |, ~) para crear expresiones más complejas y filtrar datos con mayor precisión.

```python
df.loc[(df.AtomicMass > 280)&(df.Type == "Transactinide"),:]
```

## Manipulación
La manipulación de datos es una parte esencial del análisis, que va más allá de la simple selección de filas y columnas. Una tarea común y fundamental es renombrar columnas para que los nombres sean más descriptivos, fáciles de entender y adaptados a las necesidades de tu proyecto. Esto no solo mejora la claridad de los datos, sino que también facilita el proceso de análisis. Pandas ofrece herramientas simples y eficientes para realizar esta tarea, como la función rename.

- Renombrar una columna

Si necesitas cambiar el nombre de una única columna, puedes utilizar la función rename de Pandas. Esto se hace pasando un diccionario con el nombre actual de la columna como clave y el nuevo nombre como valor. También es importante usar el argumento inplace=True para que el cambio se aplique directamente al DataFrame sin necesidad de reasignarlo a una nueva variable.

Esta es la sintaxis basica: 

```python
df.rename(columns={"AtomicNumber":"NumeroAtomico"},inplace=True)
```

## Renombrar más de una columna

Puedes renombrar múltiples columnas simultáneamente, usando un diccionario que incluya todos los cambios que deseas hacer.

Esta es la sintaxis básica:

```python
df.rename(columns={"columna1": "nuevo_nombre1", "columna2": "nuevo_nombre2"}, inplace=True)
```

## Renombrar todas las columnas

Si necesitas renombrar todas las columnas del DataFrame, puedes hacerlo asignando una lista con los nuevos nombres a la propiedad columns. Es importante asegurarte de que la lista contenga exactamente el mismo número de nombres que el número de columnas en el DataFrame.

## Duplicados
Uno de los problemas más comunes al trabajar con conjuntos de datos es la presencia de registros duplicados. Estos duplicados pueden provenir de registros completamente repetidos o de elementos específicos que aparecen varias veces con diferentes identificadores. Independientemente de su origen, la eliminación de registros duplicados es esencial para asegurar que los resultados de tu análisis sean precisos y confiables.

- Identificación de duplicados 
Pandas ofrece varias herramientas para identificar y eliminar registros duplicados, lo que permite limpiar y preparar los datos para un análisis más preciso.

Pero antes de eliminar registros duplicados, es fundamental identificarlos. Puedes utilizar la función duplicated() para verificar si las filas en un DataFrame están duplicadas. Esta función devuelve una serie booleana donde True indica que una fila es un duplicado.

```python
df.duplicated()
```

Una vez identificados los duplicados, puedes eliminarlos utilizando la función drop_duplicates(). Esta función elimina todas las filas duplicadas, manteniendo solo la primera aparición de cada duplicado.

```python
df.drop_duplicates(inplace=True)
```

## Mantener uno de los valores duplicados
Por defecto, el método drop_duplicates() mantiene la primera aparición de los datos duplicados, pero, en ocasiones, no es la que quiere mantener el programador. En este caso, emplearemos la propiedad keep de este método, que nos permite indicar si queremos conservar la primera o la última ocurrencia del duplicado. Si queremos especificar en qué columna debe buscar los duplicados Pandas, lo indicaremos con subset=["nombre_columna"].

Valores repetidos

```python
df.drop_duplicates(subset=["Discoverer"],keep="first")
```

## Reducción de las ocurrencias. 
En este caso, indicaremos a Pandas que elimine los duplicados únicamente cuando las columnas que indiquemos entre corchetes [] y separadas por comas, tengan los mismos valores en diferentes filas. Esta acción puede ser muy útil con nombres y fechas, donde es habitual mantener únicamente el último registro por considerarse el más actualizado.

```python
df.drop_duplicates(subset=["Discoverer","Year"],keep="first")
```

## Eliminación de datos
La eliminación de datos no deseados es un proceso crucial para preparar tus datos para el análisis. Pandas ofrece la función drop() para eliminar filas o columnas de manera precisa y eficiente. Dependiendo de qué datos se quieran eliminar, hay diferentes formas de utilizar esta función. 

- Filas con valores faltantes (NaN): Elimina filas que contienen valores ausentes o no válidos, asegurando que solo se analice información completa.

- Columnas irrelevantes: Elimina columnas que no aportan valor al análisis o que están duplicadas, simplificando el conjunto de datos y el proceso de análisis.

- Registros erróneos: Elimina registros que contengan errores o datos inconsistentes para garantizar la precisión de los resultados.

- El elemento axis = especifica si se quieren eliminar filas (0) o columnas (1), mientras que inplace=True modifica el Dataframe original, eliminando los datos de forma permanente. Si quieres eliminar más de una columna, ponlas entre corchetes [] y separadas por comas ,.

## Eliminar una columna o más

## Eliminar todas las filas donde haya valores faltantes (Na, NaN) en una columna

En este ejemplo se utiliza el método dropna() para eliminar todas las filas del DataFrame data que contienen valores faltantes (NaN) en la columna "Type". El parámetro subset=["Type"] indica que solo se deben considerar los valores faltantes en la columna "Type" para la eliminación de filas. El parámetro inplace=True indica que la modificación se realiza en el DataFrame original.

```python
df.dropna(subset=["Type"], inplace=True)
```
