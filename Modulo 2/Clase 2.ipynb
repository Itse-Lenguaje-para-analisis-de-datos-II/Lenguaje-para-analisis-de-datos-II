## Extract y sort_values
Los métodos extract() y sort_values() son muy útiles para seleccionar y organizar los datos necesarios para un análisis. En el caso de extract te permite buscar y extraer patrones de texto dentro de series de datos de tipo string. Por su parte, sort_values te permite ordenar los datos de un DataFrame en función de una o más columnas, permitiéndote organizar y visualizar la información de la manera que te resulte más útil.

## Ejemplos de extract
Extracción por tipo de variable (objeto, float, int...).

Este código crea un nuevo DataFrame data_object que contiene solo las columnas del DataFrame original que son de tipo objeto.

```python
df.select_dtypes(include=[object])
```

Este código crea un nuevo DataFrame data_float que contiene solo las columnas del DataFrame original que son de tipo float.

```python
df.select_dtypes(include=[float])
```

## Extracción por rango numérico 

Este código crea un nuevo DataFrame que contiene solo las filas del DataFrame original donde el valor de la columna "AtomicNumber" está entre 25 y 30 (ambos inclusive).

```python
df[df["AtomicNumber"].between(25, 30)] 
```

## Extracción por palabra específica

Este código crea un nuevo DataFrame que contiene solo las filas del DataFrame original donde la palabra "gas" se encuentra en al menos una de las columnas "Phase", "Element" o "AtomicNumber".

```python
df.loc[:,["Phase","Element","AtomicNumber"]].loc[df.Phase.str.contains("gas"),:]
```

## Extracción de la primera letra de cada palabra 

Este código crea una nueva columna con la primera letra de las palabras. La letra que extrae se indica con str[numero_de_letra].

```python
df["letra"]= df["Discoverer"].astype(str).str[0]
```

## Extracción de múltiples valores usando el método isin

Este código crea un nuevo DataFrame que contiene solo las filas del DataFrame original donde el valor de la columna "Phase" es "gas" o "solid".

```python
df.loc[:,["Phase","Symbol"]].loc[df.Phase.isin(["gas","solid"]),:]
```

## Extracción de la primera palabra de una columna

Este código crea una nueva columna con la palabra que ha identificado como la primera de cada columna. La palabra que se extrae se indica con str.get(numero_de_palabra)

```python
df["NEW"] = df.Discoverer.str.split().str.get(0)
```

## Ejemplos de sort_values
Esta función ordena el Dataframe en orden ascendente (ascending = 1) o descendente (ascending = 0) por los valores de la columna especificada a lo largo de cualquiera de los índices. Se puede aplicar a una única variable o a más de una.

```python
df.loc[:,["Year","Element","AtomicNumber"]].sort_values(by="Year",ascending =1)
```

Ascendiendo por más de una variable.

## Stats y correlaciones
Las estadísticas descriptivas y las correlaciones son herramientas esenciales para comprender las características y relaciones de tus datos. Te permiten resumir la información de manera eficiente y descubrir patrones ocultos que pueden ser cruciales para tus análisis.

## Análisis estadístico básico 
Cuando se trabaja con números, es muy probable que la estadística entre en juego en algún momento. Pandas ofrece ciertas funcionalidades para trabajar con elementos estadísticos mediante el uso de comandos. Algunos de los más usados son:

- df.max(): Calcula el valor máximo.

- df.min(): Calcula el valor mínimo.

- df.mean(): Calcula la media de los valores que se indiquen.

- df.sum(): Suma todos los valores que se indiquen.

- df.variable.value_counts(): Calcula la frecuencia de aparición de cada valor único indicado.

- df.count(): Realiza un conteo de todos los valores no nulos en cada columna.

## Correlaciones
La correlación es una medida estadística que cuantifica la fuerza y la dirección de la relación lineal entre dos variables. Esto no implica necesariamente que existe una relación causal entre las variables. Existen dos tipos de correlación: 

- Correlación positiva: Indica que dos variables se mueven en la misma dirección. Un ejemplo básico es la relación entre peso y altura, donde las personas más altas tienden a pesar más y viceversa.

- Correlación negativa: Indica que dos variables se mueven en direcciones opuestas. Un ejemplo básico es la relación entre velocidad y distancia, donde a mayor velocidad, menor tiempo se necesitará para llegar al destino y viceversa.

El valor de una correlación (coeficiente de correlación) varía entre -1 y 1 (-1: Indica una correlación negativa perfecta, 0: Indica que no existe una correlación lineal significativa, 1: Indica una correlación positiva perfecta). Por otro lado, la magnitud del valor (sin importar el signo) indica la fuerza de la relación, siendo los valores cercanos a 0 una relación débil o inexistente y los valores cercanos a 1 o -1 una relación fuerte. 




Correlación entre dos variables

```python
df.AtomicMass.corr(df.AtomicRadius)
```

Correlación de una variable con las demás

```python
df.select_dtypes(exclude=["str","object"]).corrwith(df.AtomicMass)
```

Correlación en todo el dataset

```python
df.select_dtypes(exclude=["str","object"]).corr()
```

## Group_by
Otra de las maneras de organizar los datos en Pandas es el método group_by, que permite agrupar datos en un Dataframe según los valores de una o varias columnas. Este es un concepto clave, ya que cuenta con la capacidad de resumir, agregar y agrupar datos de manera eficiente.

Este método permite crear una agrupación de categorías y aplicarles una función, que es una técnica ampliamente usada en la ciencia de datos. Un proyecto dentro de este ámbito tratará con grandes cantidades de datos y necesitará probar cosas una y otra vez, lo que puede realizarse de forma más eficiente mediante group_by.

## Ejemplos de group_by
Para todos los ejemplos, se emplea as_index_False para mantener los nombres originales de las columnas. Si no se indicara esto, las columnas perderían sus nombres y la columna afectada por el método group_by se convertiría en el index.

Agrupar por una variable y sacar la suma de todas las demás variables numéricas. Las variables numéricas que se sumarán son las creadas tras la agrupación.

```python
df.groupby(by=["Phase"],as_index=False).sum()
```

Agrupar por dos variables y sacar la media de todas las demás variables numéricas

```python
df.loc[:,["Phase","AtomicMass"]].groupby(by=["Phase"],as_index=False).mean()
```

Agrupar por tres variables y contar cuántas veces aparece cada valor en una variable

```python
df.loc[:,["Discoverer","Phase","Type"]].groupby(by= ["Discoverer","Type"],as_index=False).count()
```
