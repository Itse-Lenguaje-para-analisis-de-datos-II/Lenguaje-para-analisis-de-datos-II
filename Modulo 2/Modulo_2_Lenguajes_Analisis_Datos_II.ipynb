# Clase 2. Selección, filtrado y limpieza

**Objetivo:** Seleccionar subconjuntos por etiquetas y posiciones, aplicar condiciones booleanas y tratar datos duplicados y faltantes sin alterar inadvertidamente los originales.

**Temas:** loc e iloc; condiciones; rename; duplicated; drop; dropna; isna y fillna.

**Correspondencia con el PDF del módulo 2:** páginas 12–35.

**Lecturas para esta clase:**

- **Libro:** McKinney (2022), *Python for Data Analysis* (3.ª ed.), capítulos 5 y 7
- **Sitio web:** [Indexación](https://pandas.pydata.org/docs/user_guide/indexing.html) · [Datos faltantes](https://pandas.pydata.org/docs/user_guide/missing_data.html)

**Metodología:** explicación breve, demostración ejecutable, interpretación y práctica independiente.

### 2.1 Selección por nombre y por posición
`loc` selecciona **etiquetas** y sus extremos en rangos etiquetados pueden ser inclusivos; `iloc` selecciona **posiciones**, con extremo final excluido.

```python
df2 = df.copy()
display(df2.loc[:, ["Element", "Symbol"]].head())
display(df2.iloc[0, :])
display(df2.iloc[2:6, 1:5])
display(df2.iloc[1:3].loc[:, ["Element", "Symbol"]])
```

### 2.2 Condiciones y operadores booleanos
Para combinar comparaciones de Series utiliza `&` (y), `|` (o) y `~` (negación), encerrando cada comparación entre paréntesis.

```python
display(df2.loc[df2["AtomicMass"] > 100, ["Element", "AtomicMass"]])
display(df2.loc[(df2["AtomicMass"] > 100) & (df2["Phase"] == "solid"), ["Element", "Phase", "AtomicMass"]])
display(df2.loc[(df2["Phase"] == "gas") | (df2["Phase"] == "liquid"), ["Element", "Phase"]])
```

### 2.3 Renombrar columnas
Para conservar el original, es más claro reasignar el resultado o trabajar sobre una copia. `rename` admite varias columnas simultáneamente.

```python
renombrado = df2.rename(columns={"Element": "Elemento", "AtomicMass": "MasaAtomica"})
print(renombrado.columns.tolist())
mayusculas = df2.copy()
mayusculas.columns = mayusculas.columns.str.upper()
print(mayusculas.columns.tolist())
```

### 2.4 Duplicados
`duplicated` detecta filas repetidas, mientras `drop_duplicates` las elimina. Con `subset` se define la clave de comparación y con `keep` la ocurrencia que se conserva.

```python
duplicados_demo = pd.concat([df2.iloc[:3], df2.iloc[[1, 2]]], ignore_index=True)
print("Duplicados completos:", duplicados_demo.duplicated().sum())
display(duplicados_demo[duplicados_demo.duplicated(keep=False)])
sin_duplicados = duplicados_demo.drop_duplicates()
print("Filas luego de limpiar:", len(sin_duplicados))
display(df2.drop_duplicates(subset=["Discoverer", "Year"], keep="first").head())
```

### 2.5 Eliminar columnas o filas
`drop(columns=...)` elimina columnas seleccionadas. `dropna(subset=...)` descarta filas con valores nulos en columnas específicas. Evita eliminar columnas enteras sin evaluar su utilidad.

```python
display(df2.drop(columns=["Group", "NumberofShells"]).head(2))
print("Filas originales:", len(df2))
print("Filas con año disponible:", len(df2.dropna(subset=["Year"])))
```

### 2.6 Identificar nulos
`isna` y `isnull` son equivalentes. `sum(axis=0)` resume por columna y `sum(axis=1)` por fila.

```python
print("¿Existe algún nulo?", df2.isna().any().any())
display(df2.isna().sum().sort_values(ascending=False))
print("Total de celdas nulas:", df2.isna().sum().sum())
print("Nulos en cada una de las 5 primeras filas:")
display(df2.isna().sum(axis=1).head())
```

### 2.7 Imputación de valores faltantes
La imputación exige justificar el valor utilizado. En este ejemplo se usa la media de una variable numérica y una etiqueta explícita en una categórica. No se recomienda sustituir indiscriminadamente todos los nulos por cero.

```python
imputado = df2.copy()
imputado["AtomicRadius"] = imputado["AtomicRadius"].fillna(imputado["AtomicRadius"].mean())
imputado["Discoverer"] = imputado["Discoverer"].fillna("No informado")
print("Nulos en AtomicRadius después:", imputado["AtomicRadius"].isna().sum())
display(imputado[["Element", "AtomicRadius", "Discoverer"]].tail())
```

# Clase 3. Extracción, ordenamiento y análisis estadístico

**Objetivo:** Seleccionar variables y patrones de texto, ordenar registros, calcular estadísticas descriptivas y resumir conjuntos de datos mediante correlaciones y agrupaciones.

**Temas:** select_dtypes, between, str.contains, str.extract, isin, str.split, sort_values, describe, corr, corrwith y groupby.

**Correspondencia con el PDF del módulo 2:** páginas 36–45.

**Lecturas para esta clase:**

- **Libro:** McKinney (2022), *Python for Data Analysis* (3.ª ed.), capítulos 7 y 10
- **Sitio web:** [Datos de texto](https://pandas.pydata.org/docs/user_guide/text.html) · [Agrupación](https://pandas.pydata.org/docs/user_guide/groupby.html)

**Metodología:** explicación breve, demostración ejecutable, interpretación y práctica independiente.

### 3.1 Selección por tipo y rango
`select_dtypes` trabaja con tipos de columnas; `between` selecciona valores dentro de un intervalo inclusivo por defecto. Aunque el PDF llama «extract» a distintas selecciones, el método literal `str.extract` está destinado a extraer grupos de una expresión regular.

```python
df3 = df.copy()
display(df3.select_dtypes(include="number").head())
display(df3.select_dtypes(include="object").head())
display(df3.loc[df3["AtomicNumber"].between(25, 30), ["Element", "AtomicNumber"]])
```

### 3.2 Búsqueda y extracción de texto

```python
display(df3.loc[df3["Phase"].str.contains("gas", case=False, na=False), ["Element", "Phase"]])
df3["InicialDescubridor"] = df3["Discoverer"].astype("string").str[0]
df3["PrimeraPalabra"] = df3["Discoverer"].astype("string").str.split().str[0]
df3["CodigoExtraido"] = df3["Symbol"].astype("string").str.extract(r"^([A-Za-z]+)$", expand=False)
display(df3[["Discoverer", "InicialDescubridor", "PrimeraPalabra", "CodigoExtraido"]].head())
```

### 3.3 Selección de categorías y ordenamiento

```python
display(df3.loc[df3["Phase"].isin(["gas", "solid"]), ["Element", "Phase"]].head())
display(df3[["Year", "Element", "AtomicNumber"]].sort_values(by="Year", ascending=True, na_position="last").head())
display(df3[["Phase", "AtomicMass", "Element"]].sort_values(by=["Phase", "AtomicMass"], ascending=[True, False]).head())
```

### 3.4 Estadística descriptiva
La media y la mediana miden tendencia central; la desviación estándar describe dispersión. `describe()` ofrece una síntesis de diferentes estadísticas.

```python
numeros = df3["AtomicMass"]
print("Media:", round(numeros.mean(), 2))
print("Mediana:", round(numeros.median(), 2))
print("Desviación estándar muestral:", round(numeros.std(), 2))
display(df3[["AtomicMass", "AtomicRadius", "AtomicNumber"]].describe())
```

### 3.5 Correlación
`corr()` calcula por defecto correlación de Pearson entre variables numéricas; la correlación no demuestra causalidad. `corrwith` correlaciona columnas de un DataFrame con una Serie alineada por índice.

```python
print("Correlación masa-radio:", df3["AtomicMass"].corr(df3["AtomicRadius"]))
num = df3.select_dtypes(include="number")
display(num.corr())
display(num.corrwith(df3["AtomicMass"]).sort_values(ascending=False))
```

### 3.6 Agrupación y agregación (`groupby`)
`groupby` permite generar resúmenes por categorías. Para que la suma sea interpretable, elegimos medidas pertinentes en lugar de sumar identificadores o números atómicos.

```python
resumen_fase = df3.groupby("Phase", as_index=False).agg(
    cantidad=("Element", "count"),
    masa_media=("AtomicMass", "mean"),
    masa_maxima=("AtomicMass", "max")
)
display(resumen_fase)
display(df3.groupby(["Phase", "Type"], as_index=False).agg(
    cantidad=("Element", "size"), masa_promedio=("AtomicMass", "mean")
))
```

# Clase 4. Transformación, tablas cruzadas y combinación de DataFrames (ampliación)

**Objetivo:** Transformar valores y estructuras de tablas, construir distribuciones cruzadas y combinar datasets por concatenación o por claves relacionales.

**Temas:** replace; unique; nunique; cut; astype; to_numeric; crosstab; concat; melt; merge.

**Correspondencia con el PDF del módulo 2:** páginas ampliación basada en Clase 3(3) y Clase 3-Copy1; no desarrollado en las páginas 1–45 del PDF.

**Lecturas para esta clase:**

- **Libro:** McKinney (2022), *Python for Data Analysis* (3.ª ed.), capítulos 7 y 8
- **Sitio web:** [Reformar tablas](https://pandas.pydata.org/docs/user_guide/reshaping.html) · [Uniones](https://pandas.pydata.org/docs/user_guide/merging.html)

**Metodología:** explicación breve, demostración ejecutable, interpretación y práctica independiente.

### 4.1 Reemplazo seguro de valores
`replace` sustituye valores identificados. Usar una copia evita alterar los datos de las clases anteriores.

```python
df4 = df.copy()
df4["PhaseEsp"] = df4["Phase"].replace({"gas": "gaseoso", "solid": "sólido", "liquid": "líquido"})
df4["GroupEditado"] = df4["Group"].replace({1: 100, 18: 1800})
display(df4[["Element", "Phase", "PhaseEsp", "Group", "GroupEditado"]].head())
```

### 4.2 Valores únicos, frecuencias y tablas cruzadas
`unique` lista categorías; `nunique` cuenta categorías distintas; `crosstab` genera una tabla de contingencia. `normalize='all'` muestra proporciones del total, `index` por fila y `columns` por columna.

```python
print("Estados únicos:", df4["Phase"].unique())
print("Número de tipos:", df4["Type"].nunique())
display(df4["Phase"].value_counts())
display(pd.crosstab(df4["Phase"], df4["NumberofShells"]))
display(pd.crosstab(df4["Phase"], df4["Type"], normalize="index").round(3))
```

### 4.3 Intervalos (`cut`) y conversiones de tipo
`cut` crea grupos por rangos numéricos. Los límites deben ser consistentes con la cantidad de etiquetas. `to_numeric` permite tratar texto convertido a números.

```python
df4["CategoriaMasa"] = pd.cut(
    df4["AtomicMass"], bins=[0, 50, 150, 300],
    labels=["Ligera", "Intermedia", "Pesada"], include_lowest=True
)
texto = df4["Group"].astype("string")
recuperado = pd.to_numeric(texto, errors="coerce")
print("Tipo después de convertir:", recuperado.dtype)
display(df4[["Element", "AtomicMass", "CategoriaMasa"]])
```

### 4.4 Concatenar filas y columnas
`concat(axis=0)` apila filas, y `concat(axis=1)` agrega columnas alineando por índice. Comprueba los índices para evitar valores faltantes inesperados.

```python
a = pd.DataFrame({"Nombre": ["Carmen", "Luis"], "Edad": [22, 18]})
b = pd.DataFrame({"Nombre": ["María", "Pedro"], "Edad": [20, 25]})
filas = pd.concat([a, b], ignore_index=True)
display(filas)
sexo = pd.DataFrame({"Nombre": ["Carmen", "Luis", "María"], "Sexo": ["Mujer", "Hombre", "Mujer"]}).set_index("Nombre")
edad = pd.DataFrame({"Nombre": ["Carmen", "Luis", "María"], "Edad": [22, 18, 20]}).set_index("Nombre")
display(pd.concat([sexo, edad], axis=1))
```

### 4.5 Reestructuración con `melt`
Transforma asignaturas dispuestas en columnas (formato ancho) a filas (formato largo).

```python
notas = pd.DataFrame({
    "nombre": ["María", "Luis", "Carmen"], "edad": [18, 22, 20],
    "Matemáticas": [8.5, 7, 3.5], "Economía": [8, 6.5, 5],
    "Programación": [6.5, 4, 9]})
notas_largas = notas.melt(id_vars=["nombre", "edad"], var_name="asignatura", value_name="nota")
display(notas_largas)
```

### 4.6 `merge`: inner, left, right y outer
Los notebooks originales mostraban combinaciones con claves `id` y `name` que no existían en sus ejemplos. Aquí se definen **claves válidas** para demostrar los cuatro tipos de unión. `validate` verifica una relación muchos-a-uno cuando procede.

```python
estudiantes = pd.DataFrame({"id": [1, 2, 3], "nombre": ["Ana", "Luis", "María"]})
inscripciones = pd.DataFrame({"id": [1, 2, 4], "curso": ["Python", "SQL", "R"]})
for tipo in ["inner", "left", "right", "outer"]:
    print(f"--- MERGE {tipo.upper()} ---")
    display(pd.merge(estudiantes, inscripciones, on="id", how=tipo, validate="one_to_one"))
```

### 4.7 Caso integrador: resumen por categoría
Ejemplo combinado con `cut`, `crosstab`, `groupby` y exportación.

```python
tabla = pd.crosstab(df4["PhaseEsp"], df4["CategoriaMasa"], dropna=False)
resumen = df4.groupby("CategoriaMasa", observed=True).agg(
    elementos=("Element", "count"),
    masa_promedio=("AtomicMass", "mean")
).reset_index()
display(tabla, resumen)
resumen.to_csv(CARPETA / "reporte_integrador.csv", index=False)
```
